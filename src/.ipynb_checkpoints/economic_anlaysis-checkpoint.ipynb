- ## Segmentation of countries into developed and developing:

**The countries in this dataset are classified into two categories: Developed and Developing, based on specific thresholds for key economic indicators like GDP Growth Rate, Unemployment Rate, Inflation Rate, and Stock Index Value.**

**1. Developed Countries:**
Countries with GDP Growth Rate below 3%, Unemployment Rate below 5%, Inflation Rate below 5%, and Stock Index Value above 20,000 are considered economically stable and classified as developed.

**2. Developing Countries:**
Countries with higher GDP Growth Rates (above 3%), elevated Unemployment and Inflation Rates (above 5%), and lower Stock Index Values (below 20,000) fall under the developing category.

```python
gdp_threshold = 3
unemployment_threshold = 5
inflation_threshold = 5
stock_index_threshold = 20000

def classify_country(row):
    if (row['GDP Growth Rate (%)'] < gdp_threshold and 
        row['Unemployment Rate (%)'] < unemployment_threshold and 
        row['Inflation Rate (%)'] < inflation_threshold and 
        row['Stock Index Value'] > stock_index_threshold):
        return 'Developed'
    else:
        return 'Developing'

df['Category'] = df.apply(classify_country, axis=1)
```

### ✅ Country Classification Successful!

# 4. Data Visualization

```python
#importing the necessary modules and submodules for data visualization
import matplotlib.pyplot as plt
import seaborn as sns
```

- ## Line plots to show GDP trends over the years for multiple countries.

```python
countries = df["Country"].unique()
fig, axes = plt.subplots(len(countries), 1, figsize=(12, 5 * len(countries)), sharex=True)

for i, country in enumerate(countries):
    sns.lineplot(data=df[df["Country"] == country], x="Date", y="GDP Growth Rate (%)", marker="o", color=f"C{i % 10}", ax=axes[i])
    axes[i].set_title(f"GDP Growth Rate Trend for {country}", fontsize=14)
    axes[i].set_xlabel("Date")
    axes[i].set_ylabel("GDP Growth Rate (%)")
    axes[i].grid(True)
    axes[i].tick_params(axis="x", rotation=45)

    axes[i].xaxis.set_tick_params(labelbottom=True)

plt.tight_layout()
plt.show()
```

- ## Bar plots for the top 10 countries with the highest/lowest GDP.

- ### Aggregate the GDP data by calculating the average GDP Growth Rate for each country.

```python
avg_gdp = df.groupby(['Country'])['GDP Growth Rate (%)'].mean().reset_index()

top_5_gdp = avg_gdp.nlargest(5, 'GDP Growth Rate (%)')
bottom_5_gdp = avg_gdp.nsmallest(5, 'GDP Growth Rate (%)')
```

- ### Plotting Bar Charts:

```python
plt.figure(figsize=(15, 5))
sns.barplot(avg_gdp, x='Country', y='GDP Growth Rate (%)', hue='Country')
plt.title('GDP Growth Rate Bar Plot for all Countries', fontsize='16')
plt.xlabel('Countries', fontsize='14')
plt.ylabel('GDP Growth Rate (%)', fontsize='14')
# plt.grid(axis="x", linestyle="--", alpha=0.6)
# plt.legend(title="Countries", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

- **Top 5 countries with highest GDP:**

```python
plt.figure(figsize=(15, 5))
sns.barplot(top_5_gdp, x='Country', y='GDP Growth Rate (%)', hue='GDP Growth Rate (%)', palette='viridis', dodge=False)
plt.title('Top 5 Countries With Highest GDP Growth Rate', fontsize='16')
plt.xlabel('Countries', fontsize='14')
plt.ylabel('GDP Growth Rate (%)', fontsize='14')
# plt.grid(axis="x", linestyle="--", alpha=0.6)
plt.legend(title="GDP Growth Rate (%)", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

- **Top 5 countries with lowest GDP:**

```python
plt.figure(figsize=(15, 5))
sns.barplot(bottom_5_gdp, x='Country', y='GDP Growth Rate (%)', hue='GDP Growth Rate (%)', palette='Spectral', dodge=False)
plt.title('Top 5 Countries With Lowest GDP Growth Rate', fontsize='16')
plt.xlabel('Countries', fontsize='14')
plt.ylabel('GDP Growth Rate (%)', fontsize='14')
plt.grid(axis="x", linestyle="--", alpha=0.6)
plt.legend(title="GDP Growth Rate (%)", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
```
